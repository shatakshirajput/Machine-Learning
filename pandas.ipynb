## Series

```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

# index
print(s[0])
print(s[2])
```

```python
# label -> named index (default - 0,1,2,...)
s = pd.Series([23,24,25,26],index = ["Adam","Eve","Bob","Charlie"])
print(s)

print(s["Adam"])
print(s["Bob"])

print(s.index)

# print(s[0])    # warning will come
```

Properties of Series ->
- Homogenous.
- Vectorized Operations.
- Handle Missing Values with NaN.
- Mutable values, immutable size.

## DataFrame

```python
# Data Frame as dictionary of lists.

info = {
    "Name" : ["Adam","Eve","Bob"],
    "Age":[23,24,25],
    "CGPA":[9.5,8.4,7.6]
}

df = pd.DataFrame(info)

print(df)
print(type(df))

print(df.index)
print(df.columns)
```

```python
df
```

```python
# Data Frame as list of lists

df = pd.DataFrame([["Adam",23],["Bob",24],["Eve",25]], columns = ["Name","Age"])

print(df)
```

```python
# Data Frame as numpy arrays

import numpy as np
np_arr = np.array([[1,2,3],[4,5,6]])

df = pd.DataFrame(np_arr,columns = ["A","B","C"])
print(df)
```
