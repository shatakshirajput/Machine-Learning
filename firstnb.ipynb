```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("jsondata.json")
print(data)
print(type(data))
```

## Cleaning and structuring our data
Problems associated may be->
- data is inconsistent(mixed type)
- missing data
- duplicate data

```python
def clean_data(data):
    text_to_num ={"one": 1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data = []
    unique_data = set()
    for user in data:
        # clean rating
        raw_rating = str(user["rating"]).strip().lower()   #remove spaces(Strip)
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # missing data(Age)
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # duplicate data(filter unique values)
        if(user["name"].strip() in unique_data):
            continue
        unique_data.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

## Meaningful insights
Ques- Find ->
- avg rating
- % of users gave poor rating(<3)

```python
def get_insights(data):
    #  avg rating
    tot_rating = 0
    for user in data:
        tot_rating += float(user["rating"])
    print(f"avg rating = {tot_rating/len(data)}")

    # percentage of users with poor rating
    poor_rating = 0;
    for user in data:
        if(float(user["rating"]) < 3):
            poor_rating += 1;
    print(f"% of user with poor rating = {poor_rating/len(data) * 100} %")
```

```python
get_insights(data)
```

## Recommendation Feature
Ques-> Build a product recommendation feature(Rule based, not ML)
- if user rating >= 4, recommend same brand products.
- if user rating < 4, recommend differnent brand products.

```python
def recommendation_system(data):
    recommendations = []
    for user in data:
        curr_recom = {}
        curr_recom["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"
        recommendations.append(curr_recom)
    return recommendations
```

```python
recommendation_system(data)
```
