Q1. For the following dataset generated using NumPy:

np.random.normal(loc=70, scale=10, size=200)

- Plot a histogram of scores
- Draw a vertical line for the mean and try to interpret the results

```python
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
scores = np.random.normal(loc=70, scale=10, size=200)
```

```python
sns.histplot(scores, bins = 20,kde =True)
plt.axvline(scores.mean(),color= "red",linestyle="--",label="Mean")

plt.title("Histogram of Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

Q2. For the Seaborn penguins dataset, do the following:

- Plot flipper_length_mm vs body_mass_g
- Color by species

```python
penguins = sns.load_dataset("penguins")
```

```python
penguins.head()
```

```python
sns.scatterplot(
    data =penguins,
    x = "flipper_length_mm",
    y = "body_mass_g",
    hue = "species"
)

plt.title("Flipper Length vs Body Mass (Penguins)")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Body Mass (g)")
plt.show()
```

Q3. For the Seaborn tips dataset, do the following:
- Create a box plot of total_bill for each day
- Add hue="sex"
- Identify which day has the highest median bill

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
sns.boxplot(
    data = tips,
    x = "day",
    y= "total_bill",
    hue = "sex"
)


plt.title("Total Bill by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

Q4. For the Seaborn tips dataset, do the following:

- Create a scatter plot of total_bill vs tip
- Color points by sex
- Use different markers for smoker
- Add a title and axis labels

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker"
)

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

Q5. For the Seaborn tips dataset, do the following:

- Create a pivot table of average tip
  -- Rows: day
  -- Columns: time
- Plot the pivot table as a heatmap
- Annotate values
- Interpret which day-time combination performs best

```python
import pandas as pd

pivot = tips.pivot_table(
    values="tip",
    index="day",
    columns="time",
    aggfunc="mean",
    observed = False
)

pivot
```

```python
# Plot heatmap
sns.heatmap(pivot, annot=True, cmap="coolwarm")
plt.title("Average Tip by Day and Time")
plt.show()
```
